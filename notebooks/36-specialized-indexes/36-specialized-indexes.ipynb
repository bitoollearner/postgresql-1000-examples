````python
# Is the large dataset loaded? Run this before anything else.
import psycopg

with psycopg.connect("postgresql://book:book@localhost:5432/pg1000") as _c:
    _loaded = _c.execute(
        "SELECT count(*) FROM information_schema.schemata "
        "WHERE schema_name = 'ecommerce_lg'").fetchone()[0]
    _rows = _c.execute(
        "SELECT count(*) FROM ecommerce_lg.orders").fetchone()[0] \
        if _loaded else 0

if _loaded:
    print(f"ecommerce_lg is loaded - {_rows:,} orders. You are ready.")
else:
    print("ecommerce_lg is NOT loaded. This chapter will not work yet.")
    print("")
    print("  docker exec -it pg1000-practice load-large-dataset")
    print("")
    print("Takes about three minutes. Re-run this cell afterwards.")
````
